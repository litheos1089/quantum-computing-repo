### Step 0: Call needed Python Libraries (Including _Qiskit_)

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import Aer
from qiskit.visualization import plot_histogram, plot_bloch_vector, plot_bloch_multivector
import numpy as np
```

### Step 01: Task 1.1 - Create Superposition

Create a circuit with one qubit and apply a Hadamard gate to place the qubit into a superposition state.

Steps:
1. Create a quantum circuit with 1 qubit.
2. Apply a Hadamard gate.
3. Measure the qubit.
4. Run the simulation.
5. Display the histogram.


```python
#Create the circuit with 1 qubit (with 1 classical bit output for measurement

qc = QuantumCircuit(1,1)

# Apply Hadamard gate
qc.h(0)

# Measure
print(qc.measure_all())
```

```python
# Draw circuit
qc.draw(output="mpl", initial_state=True)
```

```python
# Run a simulation
sim = Aer.get_backend('aer_simulator')
result = sim.run(qc).result()
print(result)
```

```python
# Get counts
counts = result.get_counts()
print(counts)
```

```python
# Display the histogram
plot_histogram(counts)
```

### Step 02: Task 1.2 — Apply Multiple Quantum Gates
Modify the circuit so that it contains two different quantum gates.


```python
#Create another circuit with 1 qubit (with 1 classical bit output for measurement

qc2 = QuantumCircuit(1,1)

# Chosen Example: Apply X-Gate, followed by an H-Gate
# Apply Hadamard gate
qc2.x(0)
qc2.h(0)

# Measure
print(qc2.measure_all())
```

```python
# Draw circuit
qc2.draw(output="mpl", initial_state=True)
```

```python
# Run a simulation
sim = Aer.get_backend('aer_simulator')
result2 = sim.run(qc2).result()
print(result2)
```

```python
# Get counts
counts2 = result2.get_counts()
print(counts2)
```

```python
# Display the histogram
plot_histogram(counts2)
```
